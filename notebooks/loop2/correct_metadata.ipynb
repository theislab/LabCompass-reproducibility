# Update Metadata for New Experiments.

```python
import numpy as np
import pandas as pd
import scanpy as sc
```

## Read Updated Metadata.

```python
meta_path = "../../project_folder/data/gdrive/metadata_updated.xlsx"
meta_df = pd.read_excel(meta_path)
meta_df
```

### Get New Experiments.

```python
meta_df_lpho013 = meta_df[
    meta_df.experiment_id == "LPHO013"
]
meta_df_lpho013
```

### Map from Experiment Number to O2 and  #Days.

```python
exp_number_to_O2 = {
    int(row["experiment_number"]): row["o2_[%]"]
    for _, row in meta_df_lpho013.iterrows()
}
exp_number_to_number_of_days = {
    int(row["experiment_number"]): row["days_of_culture"]
    for _, row in meta_df_lpho013.iterrows()
}
exp_number_to_O2, exp_number_to_number_of_days
```

## Full Data.

### Read.

```python
adata_full_path = "../../project_folder/data/gdrive/loop2/BloodPlus_Loop2_logicle.h5ad"
adata_full = sc.read_h5ad(adata_full_path)
adata_full
```

### Update.

```python
meta_full = adata_full.obs.copy()

lpho013_mask = meta_full.experiment_id == "LPHO013"
meta_full_lpho013 = meta_full[
    lpho013_mask
]

assert np.all(np.isfinite(meta_full_lpho013["o2_[%]"].values))
assert np.all(np.isfinite(meta_full_lpho013["days_of_culture"].values))

meta_full_lpho013["o2_[%]"] = meta_full_lpho013.experiment_number.astype(int).map(exp_number_to_O2)
meta_full_lpho013["days_of_culture"] = meta_full_lpho013.experiment_number.astype(int).map(exp_number_to_number_of_days)

assert np.all(np.isfinite(meta_full_lpho013["o2_[%]"].values))
assert np.all(np.isfinite(meta_full_lpho013["days_of_culture"].values))

print("BEFORE: ", meta_full["o2_[%]"].unique())
print("BEFORE: ", meta_full["days_of_culture"].unique())

meta_full.loc[lpho013_mask, "o2_[%]"] = meta_full_lpho013["o2_[%]"].values
meta_full.loc[lpho013_mask, "days_of_culture"] = meta_full_lpho013["days_of_culture"].values

assert np.all(np.isfinite(meta_full["o2_[%]"].values))
assert np.all(np.isfinite(meta_full["days_of_culture"].values))

print("AFTER: ", meta_full["o2_[%]"].unique())
print("AFTER: ", meta_full["days_of_culture"].unique())

adata_full.obs = meta_full
```

### Write.

```python
adata_full_path_output = "../../project_folder/output/miscellaneous_new/meta_update_BloodPlus_Loop2_logicle.h5ad"
adata_full.write_h5ad(adata_full_path_output)
```
