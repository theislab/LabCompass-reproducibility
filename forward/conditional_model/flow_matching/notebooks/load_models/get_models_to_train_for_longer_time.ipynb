```python
import os
import yaml

import wandb
```

```python
entity = "haematopoiesis-cambridge"
project_name = "sweep_cfm_protocol_concat_X_channel_standardized+X_scatter_standardized_in_distribution_big_new_measurements"


base_config_dir = "/lustre/groups/ml01/workspace/lorenzo.consoli/projects/SFC_cambridge/collab-goettgens-SFC/forward/conditional_model/flow_matching/config"
train_config_dir = os.path.join(base_config_dir, "training")
vf_config_dir = os.path.join(base_config_dir, "vf")
```

```python
api = wandb.Api()
runs = api.runs(f"{entity}/{project_name}")
print(len(runs))
```

```python
run_names_to_relaunch = [
    "neat-wildflower-3",
    "dark-cloud-16",
    "solar-pine-15",
    "fine-gorge-23",
]
```

```python
runs_to_relaunch = [
    run for run in runs if run.name in run_names_to_relaunch
]
len(runs_to_relaunch)
```

```python
num_training_steps = 1_500_000
for run in runs_to_relaunch:
    run_name = run.name
    file_name = f"{run_name}.yaml"

    vf_cgf = run.config["vf"]
    train_cgf = run.config["training"]
    train_cgf.update({"num_training_steps": num_training_steps})
    
    train_cfg_path = os.path.join(
        train_config_dir, file_name
    )
    vf_cfg_path = os.path.join(
        vf_config_dir, file_name
    )

    with open(train_cfg_path, "w") as fb:
        yaml.dump(train_cgf, fb)

    with open(vf_cfg_path, "w") as fb:
        yaml.dump(vf_cgf, fb)
```
