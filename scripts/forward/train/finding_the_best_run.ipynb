```python
import wandb
```

```python
api = wandb.Api()

runs = api.runs("i-navosha-technical-university-of-munich/sweep_big_RA_4")
```

```python
best_run = max(
    runs,
    key=lambda r: r.summary.get("Region_weighted_f1_score", float("-inf"))
)

print("Best run:", best_run.id)
print("Best F1:", best_run.summary["Region_weighted_f1_score"])

for f in best_run.files():
    print(f.name)
```

```python
best_run.file("config.yaml").download(replace=True)
```
