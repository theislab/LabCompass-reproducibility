# Concatenate Data from Loop 3.

```python
import numpy as np
import pandas as pd
import scanpy as sc
```

## Read Loop2 Data.

```python
adata_full_loop2_path = "/lustre/groups/ml01/workspace/lorenzo.consoli/projects/SFC_cambridge/output/miscellaneous_new/concat_BloodPlus_logicle_Loop2.h5ad"
adata_full_loop2 = sc.read_h5ad(adata_full_loop2_path) 
adata_full_loop2
```

## Read Loop 3 Data.

```python
adata_full_loop3_path = "/lustre/groups/ml01/workspace/lorenzo.consoli/projects/SFC_cambridge/data/gdrive/loop3/BloodPlus_Loop3_logicle.h5ad"
adata_full_loop3 = sc.read_h5ad(adata_full_loop3_path) 
adata_full_loop3
```

### Prepare Place Holder for Original Loop 2 Data.

```python
meta_df_loop2_orig = adata_full_loop2.obs.copy()
meta_df_loop2_updated = meta_df_loop2_orig.copy()

meta_df_loop2_updated['il7_[ng_ml]'] = 0.0
meta_df_loop2_updated[ 'mcsf_[ng_ml]'] = 0.0
adata_full_loop2.obs = meta_df_loop2_updated
```

### Concatenate.

```python
adata_full_concat_loop3 = sc.concat((adata_full_loop2, adata_full_loop3), merge="same", uns_merge="same")
adata_full_concat_loop3
```

### Make Observation Names Unique.

```python
adata_full_concat_loop3.obs_names_make_unique()
```

### Update DType.

```python
adata_full_concat_loop3.obs.experiment_number = adata_full_concat_loop3.obs.experiment_number.astype(int)
adata_full_concat_loop3.obs.count_beads = adata_full_concat_loop3.obs.count_beads.astype(int)
adata_full_concat_loop3.obs.cell_counts = adata_full_concat_loop3.obs.cell_counts.astype(int)
```

### Write.

```python
adata_full_concat_path = "/lustre/groups/ml01/workspace/lorenzo.consoli/projects/SFC_cambridge/output/miscellaneous_new/concat_BloodPlus_logicle_Loop3.h5ad"
adata_full_concat_loop3.write_h5ad(adata_full_concat_path)
adata_full_concat_loop3
```
