### Import Libraries.

```python
import yaml

import numpy as np
import pandas as pd
import scanpy as sc
```

## Read Data.

### Read Annotation Configuration.

```python
annot_config_path = "/lustre/groups/ml01/workspace/lorenzo.consoli/projects/SFC_cambridge/collab-goettgens-SFC/inverse/loss_guidance/config/annotation/default.yaml"
with open(annot_config_path, "r") as fb:
    annot_config = yaml.safe_load(fb)
protocol_cols = annot_config["protocol_columns"]
protocol_cols
```

### Read Logicle Data.

```python
logicle_path = "/lustre/groups/ml01/workspace/lorenzo.consoli/projects/SFC_cambridge/data/gdrive/HID01_fcs_concatenated_logicle.h5ad"
adata_logicle = sc.read_h5ad(logicle_path)
adata_logicle
```

### Read Bloodplus Data.

```python
bloodplus_path = "/lustre/groups/ml01/workspace/lorenzo.consoli/projects/SFC_cambridge/data/gdrive/new_dataset/BloodPlus_Logicle_pm_.h5ad"
adata_bloodplus = sc.read_h5ad(bloodplus_path)
adata_bloodplus
```

## Get Unique Protocols.

### Get Unique Protocols for Old Data.

```python
logicle_protocol_df = adata_logicle.obs[protocol_cols]
for col in protocol_cols:
    logicle_protocol_df[col] = logicle_protocol_df[col].astype(float)
X_logicle_protocol = np.unique(logicle_protocol_df.values, axis=0)
print(X_logicle_protocol.shape)
```

### Get Unique Protocols for Bloodplus data.

```python
bloodplus_protocol_df = adata_bloodplus.obs[protocol_cols]
for col in protocol_cols:
    bloodplus_protocol_df[col] = bloodplus_protocol_df[col].astype(float)
X_bloodplus_protocol = np.unique(bloodplus_protocol_df.values, axis=0)
print(X_bloodplus_protocol.shape)
```

## Check overlap.

### Check overlap and differences.

```python
# Convert unique rows to set of tuples
set_logicle = set(tuple(row) for row in X_logicle_protocol)
set_bloodplus = set(tuple(row) for row in X_bloodplus_protocol)

# Overlap
common = set_logicle.intersection(set_bloodplus)
print(f"Number of unique protocols in logicle: {len(set_logicle)}")
print(f"Number of unique protocols in bloodplus: {len(set_bloodplus)}")
print(f"Number of common protocols: {len(common)}")
print(f"Fraction of logicle protocols present in bloodplus: {len(common)/len(set_logicle):.2%}")
print(f"Fraction of bloodplus protocols present in logicle: {len(common)/len(set_bloodplus):.2%}")
```

### Store the information properly.

```python
# create dataframes
df_logicle = pd.DataFrame(X_logicle_protocol, columns=protocol_cols)
df_bloodplus = pd.DataFrame(X_bloodplus_protocol, columns=protocol_cols)

# Merge with indicator
merged = pd.merge(
    df_logicle.assign(source='logicle'),
    df_bloodplus.assign(source='bloodplus'),
    on=protocol_cols,
    how='outer',
    indicator=True
)

# Split into exclusive parts
only_logicle = merged[merged['_merge'] == 'left_only'].drop(columns=['_merge', 'source_right'])
only_bloodplus = merged[merged['_merge'] == 'right_only'].drop(columns=['_merge', 'source_left'])
common = merged[merged['_merge'] == 'both'].drop(columns=['_merge', 'source_left', 'source_right'])

print(f"Only in logicle: {len(only_logicle)}")
print(f"Only in bloodplus: {len(only_bloodplus)}")
print(f"Common: {len(common)}")
```
